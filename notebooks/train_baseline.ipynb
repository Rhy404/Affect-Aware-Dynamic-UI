```python
import sys
import os
import torch
from torch.utils.data import DataLoader

# Add project root / preprocessing to python path so we can import ClipReader
sys.path.append(os.path.abspath(os.path.join('..', 'preprocessing')))

from dataset import ClipReader

# Path to configs
stats_file = os.path.abspath(os.path.join('..', 'configs', 'norm_stats.json'))

# Initialize the dataset reader
train_dataset = ClipReader(
    root=r"D:\dev\data\avcaffe",
    split="train",
    targets=["mental_demand", "effort", "temporal_demand"],
    mode="random", # Returns one random 2-second window per clip
    normalize=True,
    stats_path=stats_file
)

# Wrap in a PyTorch DataLoader
train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)

# Fetch one batch to inspect shapes
batch = next(iter(train_loader))

print("=== BATCH INSPECTION ===")
print("Video Tensor Shape:", batch["video"].shape)  # Expected: [4, 3, 16, 112, 112]
print("Audio Tensor Shape:", batch["audio"].shape)  # Expected: [4, 1, 80, 200]
print("Labels (Mental Demand, Effort, Temporal Demand):", batch["labels"])
```
